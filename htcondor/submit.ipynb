```python
import htcondor as htc
```

```python
class Job:
    def __init__(self, executable, args, id):
        self.mem = "128MB"
        self.cpus = 1
        self.id = id
        self.disk = "128MB"
        self.executable = executable
        self.args = args

    def submit(self):
        res = htc.Submit({
        "executable": f"{self.executable}",
        "arguments": f"{self.args}",                # sleep for 10 minutes
        "output": f"job-{self.id:06}.out",
        "error": f"job-{self.id:06}.err",
        "log": f"job-{self.id:06}.log",
        "request_cpus": f"{self.cpus}",
        "request_memory": f"{self.mem}",
        "request_disk": f"{self.disk}",
        })
        return res
```

```python
Njobs = 10
jobs = []
for n in range(Njobs):
    job = Job("/bin/python3", "job.py", n)
    job.cpu = "10"
    jobs.append(job)
```

```python
sched = htc.Schedd()
for j in jobs:
    submit = j.submit()
    sched.submit(submit)
```
