```python
import pandas as pd
import numpy as np
import tqdm
```

```python
with open("../results_old.csv", "r") as f:
    df = pd.read_csv(f, sep="\s+", dtype='str', keep_default_na=False)
with open("../results.csv", "r") as f:
    df2 = pd.read_csv(f, sep="\s+", dtype='str', keep_default_na=False)
```

```python
for idx, row in tqdm.tqdm(df2.iterrows(), total=len(df2)):
    h = row.hash
    sum2 = row.sum2
    row0 = df[df.hash == h].iloc[0]
    hash_0 = row0.hash
    sum2_0 = row0.sum2
    assert sum2 == sum2_0
print("All ok")
```
